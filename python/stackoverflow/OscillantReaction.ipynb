```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import integrate
```

P. 260

```python
def model(t, y, a, b):
    return np.array([
        a - y[0] - 4. * y[0] * y[1] / (1. + y[0] ** 2),
        b * y[0] * (1. - y[1] / (1. + y[0] ** 2)),
    ])
```

```python
t = np.linspace(0., 10., 200)
```

```python
solution = integrate.solve_ivp(model, [t[0], t[-1]], y0=[0.5, 0.25], t_eval=t, args=(1., 1.), method="LSODA")
```

```python
solution
```

```python
fig, axe = plt.subplots()
axe.plot(t, solution.y.T)
```

```python
x = y = np.linspace(0, 2, 200)
X, Y = np.meshgrid(x, y)
```

```python
U, V = model(None, [X, Y], 1., 1.)
```

```python
fig, axe = plt.subplots()
axe.streamplot(X, Y, U, V)
```
