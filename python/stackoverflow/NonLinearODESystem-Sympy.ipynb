```python
import sympy as sp
from sympy.abc import N, M, alpha, gamma, epsilon, mu, r, K
```

```python
f1 = gamma * N / (1 + N) - mu * M
f1
```

```python
g = sp.simplify((alpha * r * N ** 2 * (K - N) * (N - epsilon) - mu * N ** 2 - gamma * N) / mu)
```

```python
g
```

```python
f2 = f1.subs(M, g)
```

```python
n, d = sp.fraction(sp.simplify(f2))
```

```python
f3 = sp.collect(sp.expand(n), N)
f3
```

```python
parameters = [(alpha, 4 / (K - epsilon) ** 2), (K, 17.3), (epsilon, 4.9), (r, 2.7), (mu, 0.9), (gamma, 3.2)]
```

```python
f4 = f3.subs(parameters)
```

```python
sol = sp.solve(sp.Eq(f4, 0), N, dict=True)
```

```python
sol
```

```python
g.subs(parameters + [(N, sol[0][N])])
```

```python
g.subs(parameters + [(N, sol[1][N])])
```

```python
g.subs(parameters + [(N, sol[2][N])])
```
