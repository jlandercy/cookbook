```python
import sympy as sp
from sympy.abc import x, y, z
```

```python
f = x + y + z
f
```

```python
I = sp.Integral(f, (z, 0, 1-x-y), (y, 0, 1-x), (x, 0, 1))
I
```

```python
I.doit()
```

```python
I2 = sp.Integral(sp.exp(- x ** 2 - y ** 2), (x, -sp.oo, sp.oo), (y, -sp.oo, sp.oo))
```

```python
I2.doit()
```
