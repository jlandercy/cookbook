```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

```python
x = np.linspace(-6, 4, 200)
x
```

```python
p = (x - 3) * (x + 5)
p
```

```python
fig, axe = plt.subplots()
axe.plot(x, p)
axe.grid()
```

```python
x = sp.Symbol("x")
```

```python
f = (x - 3) * (x + 5)
```

```python
sp.solve(f, x)
```
