```python
x = sp.Symbol("x", real=True)
M = sp.Symbol("M", real=True, positive=True)
D = sp.Symbol("D", real=True, positive=True)
```

```python
g = sp.tan(M * x)
g
```

```python
hnum = - D * sp.sin(x)
hnum
```

```python
hdenom = 1 + D * sp.cos(x)
hdenom
```

```python
h = hnum / hdenom
h
```

```python
sp.solve(sp.Eq(g, 0), x)
```

```python
sp.solve(sp.Eq(hnum, 0), x)
```

```python
dhdx = sp.simplify(sp.diff(h, x))
```

```python
sol = sp.solve(sp.Eq(dhdx, 0), x)
```
