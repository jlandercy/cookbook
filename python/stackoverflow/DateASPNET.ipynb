```python
data = pd.DataFrame([
    {"col1": "/Date(1239018869048)/", "col2": "Lorem ipsum", "col3": 123, "col4": "/Date(1239018873049)/"},
    {"col1": "/Date(1239018870048)/", "col2": "Lorem ipsum", "col3": 345, "col4": "/Date(1239018874049)/"},
    {"col1": "/Date(1239018871048)/", "col2": "Lorem ipsum", "col3": 678, "col4": "/Date(1239018875049)/"},
    {"col1": "/Date(1239018872048)/", "col2": "Lorem ipsum", "col3": 901, "col4": "/Date(1239018876049)/"},
])
```

```python
regex = re.compile(r"\/Date\((\d+)\)\/")
```

```python
def parse_aspnet_dates(value):
    if isinstance(value, str):
        regex_match = regex.match(value)
        if regex_match:
            return pd.to_datetime(int(regex_match.groups()[0]), unit="ms")
    return value
```

```python
result = data.map(parse_aspnet_dates)
```
