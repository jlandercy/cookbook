```python
import numpy as np
from scipy import optimize
import matplotlib.pyplot as plt
```

```python
def model0(x):
    return np.round(np.abs(-0.3757609503198057 * (x - 0.2) + (0.03785161636761336)), 6)
```

```python
def model1(x):
    return np.abs(-0.3757609503198057 * (x - 0.2) + (0.03785161636761336))
```

```python
def model2(x):
    return np.power(-0.3757609503198057 * (x - 0.2) + (0.03785161636761336), 2)
```

```python
xlin = np.linspace(0, 0.5, 200)
ylin0 = model0(xlin)
ylin1 = model1(xlin)
ylin2 = model2(xlin)
```

```python
fig, axe = plt.subplots()
axe.plot(xlin, ylin0)
axe.plot(xlin, ylin1)
axe.plot(xlin, ylin2)
```

```python
optimize.minimize(model0, x0=[1], bounds=((0, np.inf),))
```

```python
optimize.minimize(model1, x0=[1], bounds=((0, np.inf),))
```

```python
optimize.minimize(model2, x0=[1], bounds=((0, np.inf),))
```
