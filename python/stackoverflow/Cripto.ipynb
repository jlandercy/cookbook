```python
import base64
import secrets
import uuid
import jwt
from cryptography.fernet import Fernet
```

```python
key = secrets.token_hex(32)
key
```

```python
payload = {
    "id": str(uuid.uuid4()),
    "email": "user@domain.com"
}
```

```python
token = jwt.encode(payload, key)
```

```python
token
```

```python
jwt.decode(token, options={"verify_signature": False})
```
