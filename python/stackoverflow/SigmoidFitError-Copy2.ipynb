```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import optimize
```

```python
def model(x, a, b, c, d):
    return a / (1. + np.exp(-c * (x - d))) + b
```

```python
np.random.seed(12345)
x = np.linspace(0, 1, 60)
p = (0.25, -0.1, 10, 0.5)
y = model(x, *p) + 0.02 * np.random.normal(size=x.size)
```

```python
popt, pcov = optimize.curve_fit(model, x, y, p0=[1,1,2,1])
popt, pcov
```

```python
xlin = np.linspace(x.min(), x.max(), 200)
ylin = model(xlin, *popt)
```

```python
fig, axe = plt.subplots()
axe.scatter(x, y)
axe.plot(xlin, ylin)
```
