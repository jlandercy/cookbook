```python
import datetime
import openpyxl
```

```python
workbook = openpyxl.load_workbook("test.xlsx")
```

```python
workbook.properties
```

```python
null_keys = ["creator", "title", "description", "subject", "identifier", "language", "lastModifiedBy", "category", "contentStatus", "version", "revision", "keywords"]
date_keys = ["created", "modified", "lastPrinted"]
for key in null_keys:
    setattr(workbook.properties, key, None)
for key in date_keys:
    setattr(workbook.properties, key, datetime.datetime(1970, 1, 1, 0, 0, 0))
```

```python
workbook.properties
```

```python
workbook.save("test.xlsx")
```
