```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import integrate
```

https://stackoverflow.com/questions/75500928/python-implementing-leapfrog-integration-for-nonlinear-spring-force-overflow

```python
def model(t, y, omega, alpha):
    return np.array([
        y[1],
        - omega ** 2 * (1 + alpha * y[0]) * y[0]
    ])
```

```python
t = np.linspace(0, 50, 2000)
```

```python
solution = integrate.solve_ivp(model, [t[0], t[-1]], y0=[1., 0.], args=(1.0, 0.5), method="LSODA", t_eval=t)
```

```python
solution
```

```python
fig, axe = plt.subplots()
axe.plot(t, solution.y.T)
axe.set_title("Dynamic System")
axe.set_xlabel("Time, $t$")
axe.set_ylabel("Variables, $y_i(t)$")
axe.legend(["Position $y(t)$", "Speed $y'(t)$"], loc="lower right")
axe.grid()
```
