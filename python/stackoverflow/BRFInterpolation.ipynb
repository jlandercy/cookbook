```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import interpolate
```

```python
n = 50
xlin = np.linspace(-3, 3, 25)
ylin = np.linspace(-3, 3, 25)
X, Y = np.meshgrid(xlin, ylin)
```

```python
def model(x, y):
    return 1. / (x ** 2 + y ** 2 + 1)
```

```python
Z = model(X, Y)
```

```python
xy = np.array([X.ravel(), Y.ravel()]).T
z = Z.ravel()
```

```python
fig, axe = plt.subplots(subplot_kw={"projection": '3d'})
axe.plot_surface(X, Y, Z)
```

```python
interpolator = interpolate.RBFInterpolator(xy, z)
```
