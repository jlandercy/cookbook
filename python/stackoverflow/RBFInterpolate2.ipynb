```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import interpolate
```

https://stackoverflow.com/questions/79724451/value-to-use-for-smoothing-for-rbfinterpolator-from-scipy-interpolate

```python
x = np.linspace(0, 1, 30).reshape(-1, 1)
y = np.sqrt(x) + 0.1 * np.random.normal(size=x.shape)
```

```python
xlin = np.linspace(-0.1, 1.1, 200).reshape(-1, 1)
```

```python
model1 = interpolate.RBFInterpolator(x, y)
yhat1 = model1(xlin)
```

```python
model2 = interpolate.RBFInterpolator(x, y, kernel="gaussian", epsilon=20)
yhat2 = model2(xlin)
```

```python
model3 = interpolate.RBFInterpolator(x, y, smoothing=1)
yhat3 = model3(xlin)
```

```python
fig, axe = plt.subplots()
axe.scatter(x, y)
axe.plot(xlin, yhat1, "--")
axe.plot(xlin, yhat2)
axe.plot(xlin, yhat3)
axe.set_ylim([-0.25, 1.25])
```
