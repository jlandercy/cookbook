```python
import numpy as np
import pandas as pd
from scipy import integrate
import matplotlib.pyplot as plt
```

```python
acceleration = pd.read_csv("Accelerometer.csv", names=["t", "x", "y", "z"], header=0).set_index("t")
dt = pd.Series(acceleration.index.diff(), index=acceleration.index, name="dt")
```

```python
acceleration.describe()
```

```python
axe = acceleration.plot()
axe.set_title("Accelerometer")
axe.set_xlabel("Time, $t$ $[s]$")
axe.set_ylabel("Acceleration, $a(t)$ $[m\cdot s^{-2}]$")
axe.grid()
```

```python
speed = acceleration.mul(dt, axis=0).cumsum()
```

```python
axe = speed.plot()
axe.set_title("Accelerometer")
axe.set_xlabel("Time, $t$ $[s]$")
axe.set_ylabel("Speed, $v(t)$ $[m\cdot s^{-1}]$")
axe.grid()
```

```python
position = speed.mul(dt, axis=0).cumsum()
```

```python
axe = position.plot()
axe.set_title("Accelerometer")
axe.set_xlabel("Time, $t$ $[s]$")
axe.set_ylabel("Position, $r(t)$ $[m]$")
axe.grid()
```

```python
fig, axe = plt.subplots(subplot_kw={"projection": "3d"})
axe.plot(*position.values.T)
axe.set_xlabel("$x$ [$m$]")
axe.set_ylabel("$y$ [$m$]")
axe.set_zlabel("$z$ [$m$]")
```
