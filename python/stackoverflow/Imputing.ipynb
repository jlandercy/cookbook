```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF
```

```python
data = pd.read_csv(
    "https://gist.githubusercontent.com/jlandercy/4efb03bf9c38e62154a84dd88ff2ebe1/raw/ee79d83b3f233077b88ce37a32e43eef6a551a1e/gistfile1.txt",
    parse_dates=["time"]
).set_index("time").drop("index", axis=1)
```

```python
original = data.copy()
```

```python
dates = data.sample(frac=0.0025, random_state=123456789).index
```

```python
for date in dates:
    data.loc[date:date + pd.Timedelta("30min"), "pressure"] = np.nan
```

```python
sample = data.dropna()["pressure"]
```

```python
X = sample.index.values.reshape(-1, 1).astype(np.int64) / 1e9 / 60 / 60 / 24
y = sample.values
```

```python
kernel = 1.0 * RBF(length_scale=1.0, length_scale_bounds=(1e-1, 1))
```

```python
gaussian_process = GaussianProcessRegressor(kernel=kernel, n_restarts_optimizer=9)
gaussian_process.fit(X, y)
```
