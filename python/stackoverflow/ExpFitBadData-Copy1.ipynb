```python
def model_1(x, a, b, c):
    return a + b * x**c


def model_2(x, d, e, f):
    return d + e * (x**f - 1) / f
```

```python
class Solver1(solvers.FitSolver1D):
    @staticmethod
    def model(x, a, b, c):
        return a + b * x[:,0] ** c
```

```python
class Solver2(solvers.FitSolver1D):
    @staticmethod
    def model(x, d, e, f):
        return d + e * (x[:,0] ** f - 1) / f
```

```python
class Solver3(solvers.FitSolver1D):
    @staticmethod
    def model(x, b, c):
        return b * x[:,0] ** c
```

```python
xlin = np.linspace(data.x0.min(), data.x0.max(), 200)
```

```python
popt1, pcov1 = optimize.curve_fit(model_1, data.x0, data.y, maxfev=5000)
```

```python
model1b = Solver1()
sol1b = model1b.fit(data, p0=popt1)
```

```python
model1 = Solver1()
sol1 = model1.fit(data, p0=[1, 1, -1])
```

```python
sol1["parameters"]
```

```python
y1lin = model1.predict(xlin.reshape(-1, 1))
```

```python
model2 = Solver2()
sol2 = model2.fit(data)
```

```python
y2lin = model2.predict(xlin.reshape(-1, 1))
```

```python
model3 = Solver3()
sol3 = model3.fit(data)
```

```python
y3lin = model3.predict(xlin.reshape(-1, 1))
```

```python
fig, axe = plt.subplots()
axe.scatter(data.x0, data.y)
axe.plot(xlin, y1lin)
axe.plot(xlin, y2lin, "--")
axe.plot(xlin, y3lin, "-.")
#axe.plot(xlin, model_1(xlin, *popt1))
axe.grid()
```
