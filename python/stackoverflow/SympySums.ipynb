```python
import sympy as sp
```

```python
x = sp.Symbol("x", positive=True)
a = sp.Symbol("a", positive=True)
r = sp.Symbol("r", positive=True)
```

```python
s = sp.Sum(1 / x ** 2, (x, 1, sp.oo))
s
```

```python
s.doit()
```

```python
s2 = sp.Sum(1 / (1 + r) ** x, (x, a, sp.oo))
s2
```

```python
s2.doit().simplify()
```
