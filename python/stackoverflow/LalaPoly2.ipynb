```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-5, 5, 200)
```

```python
fref = 4 *  x **3 - 4 * x ** 2 - 9 * x + 9
fexp = (x - 1) * (2 * x - 3) * (2 * x + 3)
```

```python
fref = 9 * x ** 2 - 30 * x + 25
fexp = (3 * x - 5) ** 2
```

```python
fref = x ** 3 - 2 * x ** 2 - 5 * x + 6
fexp = (x + 2) * (x - 1) * (x - 3)
```

```python
fig, axe = plt.subplots()
axe.plot(x, fref)
axe.plot(x, fexp, linestyle="--")
axe.grid()
```
