```python
import numpy as np
import numdifftools as nd
```

 - https://stackoverflow.com/questions/77956010/how-to-estimate-error-propagation-on-regressed-function-from-covariance-matrix-u/77964601#77964601
 - https://infoscience.epfl.ch/entities/publication/4a5b1a7e-d169-4f7f-97ad-f271b55ec91d

```python
def model(x):
    return np.array([
        x[0] ** 2 * x[1] + x[0] * x[1] ** 3,
        x[0] * x[1] + x[0] ** 2 * x[1] ** 2,
        x[0] * x[1] ** 2 - x[0] * x[1] ** 4,
])
```

```python
Cx = np.array([
    [+0.0150, -0.0025],
    [-0.0025, +0.0075],
])
```

```python
x = np.array([1., 1.])
```

```python
J = nd.Gradient(model)
```

```python
Jx = J(x)
Jx
```

```python
Cf = Jx @ Cx @ Jx.T
Cf
```

```python
sf = np.sqrt(np.diag(Cf))
sf
```
