```python
def model(t, a, rho, p, omega, phi):
    return a + rho * np.exp(- p * t) * np.sin(2 * np.pi * omega * t + phi)
```

```python
popt, pcov = optimize.curve_fit(model, t, y)
```

```python
popt, pcov
```

```python
tlin = np.linspace(t.min(), t.max(), 200)
ylin = model(tlin, *popt)
```

```python
fig, axe = plt.subplots()
axe.scatter(t, y)
axe.plot(tlin, ylin)
axe.grid()
```
