```python
import sympy as sp
from sympy.abc import N, M, alpha, gamma, epsilon, mu, r, K, rho, theta
```

 - https://stackoverflow.com/questions/49957488/convert-cartesian-expression-to-polar-expression-using-sympy
 - https://math.stackexchange.com/questions/5046462/how-many-fixed-points-this-non-linear-ode-system-has

```python
parameters = [(alpha, 4 / (K - epsilon) ** 2), (K, 17.3), (epsilon, 4.9), (r, 2.7), (mu, 0.9), (gamma, 3.2)]
```

```python
h1 = gamma * N / (1 + N) - mu * M
h1
```

```python
h2 = alpha * r * N ** 2 / (M + N) * (K - N) * (N - epsilon) - mu * N
h2
```

```python
s = sp.solve(h2, M)
s[0]
```

```python
f = sp.simplify(h1.subs(M, s[0]))
f
```

```python
n, d = sp.fraction(f)
n
```

```python
p = sp.collect(sp.expand(n), N)
p
```

```python
p0 = p.subs(parameters)
p0
```

```python
sp.solve(p0, N)
```

```python
h2p = h2.subs({M: rho * sp.sin(theta), N: rho * sp.cos(theta)})
h2p
```

```python
h2p.limit(rho, 0)
```

```python
sp.solve([
    sp.Eq(h1.subs(parameters), 0),
    sp.Eq(h2.subs(parameters), 0),
], [M, N], dict=True)
```
