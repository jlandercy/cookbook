```python
import numpy as np
import pandas as pd
from scipy import integrate, special
import matplotlib.pyplot as plt
```

```python
def upraise(t):
    return 0.2 * special.expit((0.05 * t-10.)) - 10.2 * special.expit((0.05 * t-35.))
```

```python
t = np.linspace(0, 800, 50000)
```

```python
az = upraise(t)
```

```python
fig, axe = plt.subplots()
axe.plot(t, az)
```

```python
vz = integrate.cumulative_trapezoid(az, x=t)
z = integrate.cumulative_trapezoid(vz, t[:-1])
```

```python
fig, axe = plt.subplots()
axe.plot(t[:-1], vz)
axe.plot(t[:-2], z)
axe.grid()
```
