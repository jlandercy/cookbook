```python
import numpy as np
import numdifftools as nd
import matplotlib.pyplot as plt
from scipy import special, optimize
```

```python
xlin = np.linspace(-5, 5, 2000)
ylin = np.linspace(-5, 5, 2000)
X, Y = np.meshgrid(xlin, ylin)
```

```python
Z = optimize.rosen([X, Y])
```

```python
solution = optimize.minimize(optimize.rosen, x0=[0, 0])
solution
```

```python
fig, axe = plt.subplots()
axe.contour(X, Y, np.log10(Z), 20)
axe.scatter(*solution.x)
axe.grid()
```

```python
gradient = nd.Gradient(optimize.rosen)
hessian = nd.Hessian(optimize.rosen)
```
