```python
import numpy as np
from scipy import integrate
import matplotlib.pyplot as plt
```

```python
def model(t, y, alpha, beta, gamma, delta):
    return np.array([
        + alpha * y[0] - beta * y[0] * y[1]
        - gamma * y[1] + delta * y[0] * y[1]
    ])
```

```python
np.random.seed(123456)
```

```python
n = 50

t = np.linspace(0, 100, 500)

p0 = np.array([
    np.random.uniform(0.5, 2.0, size=n),
    np.random.uniform(0.5, 2.0, size=n),
    np.random.uniform(0.5, 2.0, size=n),
    np.random.uniform(0.5, 2.0, size=n),
]).T

y0 = np.array([
    np.random.uniform(1., 50., size=n),
    np.random.uniform(1., 50., size=n),
]).T
```

```python
solutions = []
for i in range(n):
    solution = integrate.solve_ivp(model, [t.min(), t.max()], y0=y0[i,:], args=p0[i,:], t_eval=t, method="DOP853")
    print(solution.message)
    #print(p0[i,:])
```
