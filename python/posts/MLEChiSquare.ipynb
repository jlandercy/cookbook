```python
import sympy as sp
```

```python
x = sp.Symbol("x")
mu = sp.Symbol("mu", real=True)
sigma = sp.Symbol("sigma", real=True, positive=True)
i = sp.Symbol("i", integral=True, positive=True)
n = sp.Symbol("n", integral=True, positive=True)
xi = sp.Indexed(x, i)
mui = sp.Indexed(mu, i)
sigmai = sp.Indexed(sigma, i)
```

```python
likelihood = sp.Product(sp.exp(-((xi - mui) / sigmai) ** 2), (i,1, n))
likelihood
```

```python
loglikelihood = sp.expand_log(sp.log(likelihood), force=True)
loglikelihood
```
